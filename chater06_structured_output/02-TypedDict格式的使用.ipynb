## 2、基本使用

举例1：

deepseek大模型

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
import os

load_dotenv(verbose=True)

model_deepseek = init_chat_model(
    model="deepseek-v4-pro",
    model_provider="deepseek",
    api_key = os.getenv("DEEPSEEK_API_KEY"),
    base_url = os.getenv("DEEPSEEK_BASE_URL"),
    extra_body = {
        "thinking":{
            "type":"disabled"
        }
    }
)
```

千问大模型

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
import os

load_dotenv(verbose=True)

model_qwen = init_chat_model(
    model="qwen3.8-max",
    model_provider="openai",
    api_key = os.getenv("DASHSCOPE_API_KEY"),
    base_url = os.getenv("DASHSCOPE_BASE_URL"),
    extra_body = {
        "thinking":{
            "type":"disabled"
        }
    }
)
```

返回简单结构

```python
from typing_extensions import Annotated, TypedDict

class MovieTypedDict(TypedDict):
    """电影的信息"""
    title: Annotated[str,"电影的名称"]
    year: Annotated[int,"上映的时间,四位数"]
    director: Annotated[str,"电影的导演"]
    rating:Annotated[float, "电影的评分，满分是十分，可以包含一位小数"]


structured_model = model_qwen.with_structured_output(MovieTypedDict)
result = structured_model.invoke("给我介绍一下电影 《星际穿越》")
print(result)
print(type(result))
```

举例2：嵌套结构的使用

```python
from typing import List

class Actor(TypedDict):
    """演员的信息"""
    name : Annotated[str,"演员的姓名"]
    role : Annotated[str,"扮演的角色"]

class MovieTypedDict(TypedDict):
    """电影的信息"""
    title: Annotated[str,"电影的名称"]
    year: Annotated[int,"上映的时间,四位数"]
    director: Annotated[str,"电影的导演"]
    rating:Annotated[float, "电影的评分，满分是十分，可以包含一位小数"]
    cast:Annotated[List[Actor],"演员列表"]  #嵌套列表的结构

structured_model1 = model_qwen.with_structured_output(MovieTypedDict)
response = structured_model1.invoke("给我介绍一下电影《星际穿越》")
print(response)
print(type(response))
```

举例3：...的使用

不同的模型提供商对应的效果不同

先以DeepSeek模型为例

```python
class MoveDict(TypedDict):
    """电影的信息"""
    title : Annotated[str,...,"电影的名称"]
    year : Annotated[int,...,"上映的时间，四位数"]
    director : Annotated[str,...,"电影的导演"]
    rating : Annotated[float,...,"电影的评分，满分是十分，可以包含一位小数"]

structured_model = model_deepseek.with_structured_output(MoveDict)
response = structured_model.invoke("根据这段话抽取盗梦空间的信息，不包含的信息可以留空：盗梦空间在2010年上映，导演是克里斯托弗·诺兰。")
print(response)
print(type(response))
```
