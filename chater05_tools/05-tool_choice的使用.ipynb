# tool_choice的使用

## 1、tool_choice的取值为none时

举例：

```python
from langchain_core.tools import tool
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
import os
from rich import print as rprint

load_dotenv(override=True)
model = init_chat_model(
    model="kimi-k2.6",
    model_provider="openai",
    api_key=os.getenv("MOONSHOT_API_KEY"),
    base_url=os.getenv("MOONSHOT_BASE_URL"),
    extra_body={
        "thinking": {
            "type": "disabled"
        }
    }

)

@tool(parse_docstring=True)
def get_weather(city:str):
    """
    获取当日的天气

    Args:
        city:城市名称
    """
    return f"{city}天气晴朗"

model_with_tools = model.bind_tools([get_weather],tool_choice="none")

response = model_with_tools.invoke("江西南昌今天的天气如何?")
rprint(response)
```

## 3、tool_choice取值为required时

举例：

```python
from langchain_core.tools import tool
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
import os
from rich import print as rprint

load_dotenv(override=True)
model = init_chat_model(
    model="kimi-k2.6",
    model_provider="openai",
    api_key=os.getenv("MOONSHOT_API_KEY"),
    base_url=os.getenv("MOONSHOT_BASE_URL"),
    #因为后面设置的是强制模型使用工具required，但是Kimi K2.6模型的thinking模型不允许强制调用，所以这里需要关掉thinking
    extra_body={
        "thinking": {
            "type": "disabled"
        }
    }
)

@tool(parse_docstring=True)
def get_weather(city:str):
    """
    获取当日的天气

    Args:
        city:城市名称
    """
    return f"{city}天气晴朗"

model_with_tools = model.bind_tools([get_weather],tool_choice="required")

response = model_with_tools.invoke("江西南昌今天的天气如何?")
rprint(response)
```

## 强制调用指定的工具

举例：

```python
from langchain_core.tools import tool
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
import os
from rich import print as rprint

load_dotenv(override=True)
model = init_chat_model(
    model="kimi-k2.6",
    model_provider="openai",
    api_key=os.getenv("MOONSHOT_API_KEY"),
    base_url=os.getenv("MOONSHOT_BASE_URL"),
    #因为后面设置的是强制模型使用工具required，但是Kimi K2.6模型的thinking模型不允许强制调用，所以这里需要关掉thinking
    extra_body={
        "thinking": {
            "type": "disabled"
        }
    }
)

@tool(parse_docstring=True)
def get_weather1(city:str):
    """
    获取当日的天气

    Args:
        city:城市名称
    """
    return f"{city}天气晴朗"

@tool()
def get_weather2(city :str):
    """
    获取当日的天气

    Args:
        city:城市名称
    """
    return f"{city}天气晴朗"

model_with_tools = model.bind_tools([get_weather1,get_weather2],tool_choice="get_weather2")


response = model_with_tools.invoke("南昌今天的天气怎么样?")
rprint(response)
```
