## 1.ChatPromptTemplate的实例化

### 1.1调用from_messages()方法  --推荐

举例：

```python
from langchain_core.prompts import ChatPromptTemplate


#这样就是通过from_messages进行的一个实例化
chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的AI助手，你的名字叫{name}"),
    ("human","你好，最近怎么样？"),
    ("ai","我很好，谢谢"),
    ("human","{user_input}")
])

#实例完成后，下面就是调用
result = chat_prompt_template.invoke({"name":"小智","user_input":"9 + 9 = ？"})
print(result)
```

## 2.ChatPrompTemplate的调用

### 2.1 invoke()

传入参数：字典列表；返回值类型：ChatPromptValue


举例：

```python
from langchain_core.prompts import ChatPromptTemplate


#这样就是通过from_messages进行的一个实例化
chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的AI助手，你的名字叫{name}"),
    ("human","你好，最近怎么样？"),
    ("ai","我很好，谢谢"),
    ("human","{user_input}")
])

#实例完成后，下面就是调用
result = chat_prompt_template.invoke({"name":"小智","user_input":"9 + 9 = ？"})
print(result)
print(type(result))
```

### 2.2 format()

参数类型：变量值；返回值类型：字符串

举例：

```python
from langchain_core.prompts import ChatPromptTemplate


#这样就是通过from_messages进行的一个实例化
chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的AI助手，你的名字叫{name}"),
    ("human","你好，最近怎么样？"),
    ("ai","我很好，谢谢"),
    ("human","{user_input}")
])

#实例完成后，下面就是调用
result = chat_prompt_template.format(name="小智",user_input="9 + 9 = ?")
print(result)
print(type(result))
```

### 2.3 format_messages()

传入参数：变量值；返回值类型：消息列表

举例：

```python
from langchain_core.prompts import ChatPromptTemplate


#这样就是通过from_messages进行的一个实例化
chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的AI助手，你的名字叫{name}"),
    ("human","你好，最近怎么样？"),
    ("ai","我很好，谢谢"),
    ("human","{user_input}")
])

#实例完成后，下面就是调用
result = chat_prompt_template.format_messages(name="小智",user_input="9 + 9 = ？")
print(result)
print(type(result))
```

## 3.结合大模型的调用

举例：

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
import os
from langchain_core.prompts import ChatPromptTemplate

#####1、提供大模型#####
load_dotenv(override=True)

model = init_chat_model(
    model="kimi-k2.6",
    model_provider="openai",
    api_key=os.getenv("MOONSHOT_API_KEY"),
    base_url=os.getenv("MOONSHOT_BASE_URL"),
)

#####2、提供提示词模板#####
chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的AI助手，你的名字叫{name}"),
    ("human","你好，最近怎么样"),
    ("ai","我很好，谢谢"),
    ("human","{user_input}")
])

prompt_value = chat_prompt_template.invoke({"name":"小智","user_input":"9 + 9 = ?"})

#####3、模型调用#####
response = model.invoke(prompt_value)
print(response)
```

## 4.更丰富的初始化参数类型

### 4.1 字符串列表

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate.from_messages([
    "你好，我是{name}"    #结果会理解为一个用户消息
])

chat_prompt_template.invoke({"name":"小明"})
```

### 4.2 元组列表

```python
from langchain_core.prompts import ChatPromptTemplate
chat_prompt_template2 = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的AI助手"),
    ("human","你好，我是{name}")
])

chat_prompt_template2.invoke({"name":"小明"})
```

### 4.3字典列表

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template3 = ChatPromptTemplate.from_messages([
    {"role":"system","content":"你是一个友好的AI助手"},
    {"role":"user","content":"你好，我是{name}"}
])

chat_prompt_template3.invoke({"name":"小明"})
```

### 4.5BaseMessagePromptTemplate参数列表

```python
from langchain_core.prompts import SystemMessagePromptTemplate,HumanMessagePromptTemplate
from langchain_core.prompts import ChatPromptTemplate

system_message_template = SystemMessagePromptTemplate.from_template("你是一个友好的AI助手")
human_message_template = HumanMessagePromptTemplate.from_template("你好,我是{name}")

chat_prompt_template6 = ChatPromptTemplate.from_messages([
    system_message_template,
    human_message_template
])

chat_prompt_template6.invoke({"name":"小明"})
```

### 4.6BaseChatPromptTemplate参数列表

```python
inner_chat_template1 = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的AI助手")
])
inner_chat_template2 = ChatPromptTemplate.from_messages([
    ("human","你好,我是{name}")
])

chat_prompt_template7 = ChatPromptTemplate.from_messages([
    inner_chat_template1,
    inner_chat_template2
])

chat_prompt_template7.invoke({"name":"小明"})
```

举例：综合使用

```python
from langchain_core.messages import SystemMessage,HumanMessage
from langchain_core.prompts import ChatPromptTemplate,HumanMessagePromptTemplate,SystemMessagePromptTemplate

system_msg = SystemMessage(content="你是一个AI工程师")
human_msg = HumanMessage(content="你好!")

system_prompt = SystemMessagePromptTemplate.from_template("你是一个{role}")
human_prompt = HumanMessagePromptTemplate.from_template("{user_input}")

nested_prompt = ChatPromptTemplate.from_messages([("system","嵌套提示词")])

prompt = ChatPromptTemplate.from_messages([
    system_msg,
    human_msg,
    system_prompt,
    human_prompt,
    nested_prompt
])

prompt.invoke({"role":"人工智能专家","user_input":"介绍一下大模型的应用场景"})
```
